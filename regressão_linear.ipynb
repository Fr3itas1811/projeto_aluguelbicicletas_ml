```python
from sklearn.linear_model import LinearRegression


# Orçamento de Marketing
x = [[1000],[2500],[3250],[5000],[6500]]
# Lucro em Vendas
y = [10_000, 20_500, 33_250, 59_000, 77_000]


modelo = LinearRegression()


# Treinando o modelo
modelo.fit(x,y)


predicao = modelo.predict([[5000]])


print(f"Predição de lucro: R$ {predicao[0]:,.2f}")



```

```python
from sklearn.linear_model import LinearRegression

from sklearn.linear_model import LinearRegression

X_treino = [[10], [15], [20], [25], [30], [35], [40], [50], [55], [60]]

y_treino = [10.00, 13.00, 14.00, 18.00, 19.00, 23.00, 24.00, 30.00, 32.00, 34.00]

modelo = LinearRegression()

modelo.fit(X_treino, y_treino)

preco_estimado_45 = modelo.predict([[45]])[0]
preco_estimado_20 = modelo.predict([[20]])[0]
preco_estimado_60 = modelo.predict([[60]])[0]

print(f"Preço estimado para 45 minutos: R$ {preco_estimado_45:.2f}")
print(f"Preço estimado para 20 minutos: R$ {preco_estimado_20:.2f}")
print(f"Preço estimado para 60 minutos: R$ {preco_estimado_60:.2f}")
```
